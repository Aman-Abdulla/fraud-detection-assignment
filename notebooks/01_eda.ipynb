# 01 - Exploratory Data Analysis

Goal: understand the fraud dataset, check class balance, inspect feature distributions, and look for obvious leakage before modeling.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)
```

```python
df = pd.read_csv("../data/raw/credit_card_fraud_10k.csv")
df.head()
```

```python
df.info()
print("\nShape:", df.shape)
print("\nMissing values:\n", df.isnull().sum())
```

```python
# Class balance - critical for imbalanced fraud detection
print(df["is_fraud"].value_counts())
print("\nFraud rate: {:.2f}%\n".format(100 * df["is_fraud"].mean()))
```

```python
# Numeric feature distributions by class
numeric_cols = ["amount", "transaction_hour", "device_trust_score", "velocity_last_24h", "cardholder_age"]
df[numeric_cols].describe()
```

```python
# Boxplots: numeric features vs fraud label
fig, axes = plt.subplots(2, 3, figsize=(15, 8))
axes = axes.ravel()
for i, col in enumerate(numeric_cols):
    sns.boxplot(x="is_fraud", y=col, data=df, ax=axes[i])
    axes[i].set_title(f"{col} by is_fraud")
plt.tight_layout()
plt.show()
```

```python
# Categorical features
cat_cols = ["merchant_category", "foreign_transaction", "location_mismatch"]
for col in cat_cols:
    print(f"\n=== {col} ===")
    print(pd.crosstab(df[col], df["is_fraud"], normalize="index").round(3))
```

```python
# Correlation heatmap (numeric only)
corr = df[numeric_cols + ["is_fraud"]].corr()
sns.heatmap(corr, annot=True, cmap="RdBu_r", center=0, fmt=".2f")
plt.title("Correlation heatmap")
plt.show()
```

## Key takeaways
- Note the fraud rate (imbalanced → use PR-AUC, not accuracy)
- Check for features with near-perfect correlation to `is_fraud` (leakage)
- Identify which numeric features separate fraud vs non-fraud well
- Categorical features: which categories have higher fraud rates?
